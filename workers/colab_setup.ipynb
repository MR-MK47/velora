```python
import subprocess, sys, os
# Install static ffmpeg 7.x build (supports modern filters; apt-get provides ffmpeg 4.x)
FFMPEG_URL = 'https://github.com/BtbN/FFmpeg-Builds/releases/download/latest/ffmpeg-master-latest-linux64-gpl.tar.xz'
subprocess.run(['wget', '-q', FFMPEG_URL, '-O', '/tmp/ffmpeg.tar.xz'], check=True)
subprocess.run(['tar', '-xf', '/tmp/ffmpeg.tar.xz', '--strip=2', '-C', '/usr/local/',
                'ffmpeg-master-latest-linux64-gpl/bin/ffmpeg',
                'ffmpeg-master-latest-linux64-gpl/bin/ffprobe'], check=True)
os.remove('/tmp/ffmpeg.tar.xz')
subprocess.run(['ffmpeg', '-version'], check=True)
print('ffmpeg installed ok')
```

```python
import sys, os, subprocess
# Clone latest worker code from GitHub (always in sync with repo)
REPO_URL = 'https://github.com/MR-MK47/velora.git'
REPO_DIR = '/content/velora'
if os.path.exists(REPO_DIR):
    subprocess.run(['git', '-C', REPO_DIR, 'pull'], check=True)
else:
    subprocess.run(['git', 'clone', REPO_URL, REPO_DIR], check=True)
sys.path.insert(0, REPO_DIR)
```
